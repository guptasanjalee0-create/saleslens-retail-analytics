## 1. Import libraries and project functions

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from src.config import ROOT_DIR, CHART_DIR
from src.analysis import load_source_data, prepare_data, make_summary, build_rfm
```

## 2. Load the data

```python
raw, source = load_source_data(ROOT_DIR)
print('Source:', source)
print('Shape:', raw.shape)
display(raw.head())
```

## 3. Explore and clean the data

```python
raw.info()
raw.isna().sum()
```

```python
sales, quality = prepare_data(raw)
print('Cleaning summary:', quality)
display(sales.head())
```

## 4. Calculate key business metrics

```python
summary = make_summary(sales)
summary
```

## 5. Monthly revenue and top products

```python
monthly = sales.groupby('InvoiceMonth')['Revenue'].sum().sort_index()
monthly.plot(figsize=(10,4), marker='o', title='Monthly Revenue')
plt.ylabel('Revenue (£)'); plt.grid(alpha=.25); plt.show()
```

```python
top_products = sales.groupby('Description')['Revenue'].sum().nlargest(10).sort_values()
top_products.plot(kind='barh', figsize=(9,5), title='Top 10 Products by Revenue')
plt.xlabel('Revenue (£)'); plt.show()
```

## 6. Customer RFM analysis

```python
rfm = build_rfm(sales)
display(rfm.head(10))
rfm['Segment'].value_counts()
```
